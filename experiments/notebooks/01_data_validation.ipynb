Validate the Parquet tables before strategy research.

```python
from pathlib import Path
import polars as pl

DATASET = Path.cwd().parents[1] / 'experiments/data/parquet/btc-capture-001'
if not (DATASET / 'dataset-metadata.json').exists():
    DATASET = Path('/Users/pavelkushneryk/Documents/vsprojects/cross-venue-arbitrage/experiments/data/parquet/btc-capture-001')

for kind in ['order_book_events', 'order_book_levels', 'market_trades', 'best_bid_offers']:
    files = list(DATASET.glob(f'**/event_type={kind}/**/*.parquet'))
    lf = pl.scan_parquet([str(p) for p in files], hive_partitioning=True)
    print(kind, lf.select(pl.len()).collect())
```
