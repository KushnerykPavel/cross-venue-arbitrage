# Data validation
Validate the Parquet tables before strategy research.

```python
from pathlib import Path
import polars as pl

DATASET = Path.cwd().parents[1] / 'experiments/data/parquet/6b556515-5156-4181-8a6f-7403fa9078fe'

for kind in ['order_book_events', 'order_book_levels', 'market_trades']:
    files = list(DATASET.glob(f'**/event_type={kind}/**/*.parquet'))
    lf = pl.scan_parquet([str(p) for p in files], hive_partitioning=True)
    print(kind, lf.select(pl.len()).collect())
```
